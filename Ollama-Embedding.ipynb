```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
def genrate_response(user_query:str):

    response=vector_store.similarity_search(user_query)
    context="".join([doc.page_content for doc in response])

    prompt=PromptTemplate(template="""you are a helpful assistent of encient indian Ramayana Mythology expert.
                                    answer the following question based only on the provided context.
                                     if the inforat is not enough in the provided context to answer the question then simply answer : I do not have enough information right now. 
                                     
                                     CONTEXT : {context} \n
                                     QUESTION : {user_query} \n
                                     ANSWER : 
                                     """,
                            input_variables=['context','user_query'])

    llm=ChatOllama(model='deepseek-r1:8b')
    parser=StrOutputParser()

    chain = prompt | llm | parser

    final_response=chain.invoke({'context':context,'user_query':user_query})
    return final_response
```

```python
response=genrate_response('summarize me 2 lines who is ravana.')
```
