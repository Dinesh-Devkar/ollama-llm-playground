```python
# stream=requests.post(url=url,json=payload)
# for chunk in stream:
#     print(chunk)

import requests
import json

url = "http://localhost:11434/api/generate"

payload = {
    "model": "gemma3:4b",
    "prompt": "Explain artificial intelligence in simple terms.",
    "stream": True
}

with requests.post(url, json=payload,stream=True) as response:
    response.raise_for_status()

    for line in response.iter_lines():
        if line:
            chunk = json.loads(line.decode("utf-8"))

            # Print each piece of the generated response
            print(chunk.get("response", ""), end="", flush=True)

print()
```

## simple text summarization method

```python
def summarize(prompt:str,model:str='gemma3:4b'):
    url='http://localhost:11434/api/generate'
    payload={'model':model,'prompt':prompt,'system':'you are a helpful assistent who always summarize the give input prompt text'}

    response=requests.post(url=url,json=payload)
    response.raise_for_status()
    output=''
    for i in response.iter_lines():
        i=json.loads(i)
        output+=i['response']
    return output
```
