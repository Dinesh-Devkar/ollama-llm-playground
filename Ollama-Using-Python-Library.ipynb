```python
import ollama
```

```python
### Generate Function
```

```python
response=ollama.generate(model='gemma:7b',prompt='what is the capital of nepal',)
```

```python
print(response)
```

```python
### stream parameter
```

```python
response=ollama.generate(model='gemma:7b',prompt='who is virat kohli',stream=True)
for i in response:
    print(i['response'], end="")
```

```python
### system parameter
```

```python
response=ollama.generate(model='gemma:7b',prompt='who is virat kohli, explain in 2-3 lines.',stream=True,system="you are a funny/comedy assistent")
for i in response:
    print(i['response'], end="")
```

```python
### images parameter
```

```python
import base64

image_path = "./images/img_1.jpg"

with open(image_path, "rb") as image_file:
    image_base64 = base64.b64encode(image_file.read()).decode("utf-8")

response=ollama.generate(model="gemma3:4b",prompt="summarize me the following image",images=[image_base64],stream=True)
for i in response:
    print(i['response'], end="")
```
