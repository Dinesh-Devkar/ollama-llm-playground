### images parameter

```python
import base64

image_path = "./images/img_1.jpg"

with open(image_path, "rb") as image_file:
    image_base64 = base64.b64encode(image_file.read()).decode("utf-8")

response=ollama.generate(model="gemma3:4b",prompt="summarize me the following image",images=[image_base64],stream=True)
for i in response:
    print(i['response'], end="")
```

### format parameter with pydantic

```python
from pydantic import BaseModel,Field
```

```python
class Person(BaseModel):
    name:str=Field(description='name of the person')
    age:int=Field(description='age of the person',gt=30,lt=32)
    hobbies:list=Field(description='hobbies of the person')
```

```python
response=ollama.generate(model='gemma:7b',prompt="give me some information of one dummy person",format=Person.model_json_schema())
```

```python
print(response)
```

### think parameter

```python

from ollama import generate

stream = generate(
    model="olmo-3:7b",
    prompt="Explain step by step why 0.1 + 0.2 != 0.3 in Python.",
    think='low',
    stream=True
)

thinking_started = False
answer_started = False

for chunk in stream:

    # Display model thinking in real time
    if chunk.thinking:
        if not thinking_started:
            print("🧠 MODEL THINKING:\n")
            thinking_started = True

        print(chunk.thinking, end="", flush=True)

    # Display final answer in real time
    if chunk.response:
        if not answer_started:
            print("\n\n✅ FINAL ANSWER:\n")
            answer_started = True

        print(chunk.response, end="", flush=True)
```
